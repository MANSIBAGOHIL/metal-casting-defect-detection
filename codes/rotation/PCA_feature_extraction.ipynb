```python
import os
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from skimage.io import imread
from skimage.util import img_as_float
```

```python
# Path to your grayscale image dataset folder
dataset_path = r'D:\Capston Project\Augmentation\test'

# Initialize lists to store image data, names, and labels
image_data = []
image_names = []
labels = []
```

```python
for filename in os.listdir(dataset_path):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):  # Check for valid image types
        file_path = os.path.join(dataset_path, filename)
        
        # Check if the filename starts with "perfect" for labeling
        if filename.lower().startswith('perfect'):
            labels.append(1)  # Label as '1' for perfect
        else:
            labels.append(0)  # Label as '0' for non-perfect
        
        # Load and flatten the grayscale image
        image = img_as_float(imread(file_path, as_gray=True))  # Load grayscale image
        image_data.append(image.flatten())  # Flatten 500x500 image into a single row
        image_names.append(filename)
```

```python
# Convert to numpy array
image_data = np.array(image_data)

# Apply PCA
n_components = 50  # Adjust the number of components as needed
pca = PCA(n_components=n_components)
pca_features = pca.fit_transform(image_data)
```

```python
# Save PCA features to a CSV file
pca_df = pd.DataFrame(pca_features, columns=[f'PC{i+1}' for i in range(n_components)])
pca_df['Filename'] = image_names  # Add filenames for reference
pca_df['Label'] = labels  # Add labels for each image
pca_csv_path = r'D:\Capston Project\Augmentation\PCA_Features_With_Labels_test.csv'
pca_df.to_csv(pca_csv_path, index=False)

print(f'PCA features with labels saved to {pca_csv_path}')
```
